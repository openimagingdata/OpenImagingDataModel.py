```python
import lancedb
import openai
from dotenv import dotenv_values

env = dotenv_values()
openai.api_key = env["OPENAI_API_KEY"]
```

```python
db_conn = await lancedb.connect_async(uri=env["LANCEDB_URI"], api_key=env["LANCEDB_API_KEY"])
```

```python
tables = await db_conn.table_names()
print("Available tables:")
for table in tables:
    print(f" - {table}")
```

```python
snomedct = await db_conn.open_table("snomedct")
print("Table schema:")
schema = await snomedct.schema()
print(schema)
```

```python
await snomedct.count_rows()
```

```python
pcl_search_query = await snomedct.search("pcl tear", query_type="hybrid", vector_column_name="vector")
pcl_search_results = [(r["concept_id"], r["concept_text"], r["_relevance_score"]) for r in await pcl_search_query.to_list()]
```
